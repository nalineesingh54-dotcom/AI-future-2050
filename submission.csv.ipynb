## 🧠 Brain Tumor Detection using CNN

This project focuses on detecting and classifying brain tumors from MRI images using deep learning techniques.

### 📌 Problem Statement
Brain tumors are life-threatening conditions that require early and accurate diagnosis. Manual analysis of MRI scans can be time-consuming and error-prone.

### 🎯 Objective
To build an automated system that can classify MRI images into four categories:
- Glioma Tumor
- Meningioma Tumor
- Pituitary Tumor
- No Tumor

### ⚙️ Approach
- Image preprocessing (resizing and normalization)
- Data augmentation using ImageDataGenerator
- CNN (Convolutional Neural Network) model for classification
- Model training and validation

### 🧪 Model Architecture
- Conv2D layers
- MaxPooling layers
- Dense layers
- Softmax output layer (4 classes)

### 📊 Evaluation Metrics
- Accuracy
- Precision
- F1 Score
- Confusion Matrix

### 📁 Output
The model generates predictions for test images and saves them in a CSV file (submission.csv) for Kaggle evaluation.

### 🚀 Result
The model successfully classifies MRI images and provides a baseline solution for brain tumor detection.
